## Setup

```python
import time
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.inspection import permutation_importance, PartialDependenceDisplay
from sklearn.metrics import roc_auc_score, accuracy_score
import lightgbm as lgb
warnings.filterwarnings("ignore", message="IProgress not found")
warnings.filterwarnings("ignore", message="LightGBM binary classifier with TreeExplainer")
import shap
from lime.lime_tabular import LimeTabularExplainer

adult = pd.read_csv("data/adult.csv").replace("?", "Unknown")
adult["high_income"] = (adult["income"] == ">50K").astype(int)
CATEGORICAL = ["workclass", "marital-status", "occupation", "relationship", "race", "gender", "native-country"]
NUMERIC = ["age", "educational-num", "capital-gain", "capital-loss", "hours-per-week"]
FEATURES = NUMERIC + CATEGORICAL

# Integer codes for the categories, so every explainer can work on a plain numeric table
CATEGORIES = {c: sorted(adult[c].unique()) for c in CATEGORICAL}
X = adult[FEATURES].copy()
X[NUMERIC] = X[NUMERIC].astype(float)
for c in CATEGORICAL:
    X[c] = X[c].map({v: i for i, v in enumerate(CATEGORIES[c])}).astype(int)
y = adult["high_income"].values
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, stratify=y, random_state=42)

def train_model(X_tr, y_tr, features):
    cats = [c for c in CATEGORICAL if c in features]
    return lgb.LGBMClassifier(n_estimators=400, learning_rate=0.05, num_leaves=31, subsample=0.8, subsample_freq=1,
                              colsample_bytree=0.8, random_state=42, n_jobs=2, verbose=-1).fit(
        X_tr[features], y_tr, categorical_feature=cats)

model = train_model(X_train, y_train, FEATURES)
p_test = model.predict_proba(X_test)[:, 1]
print(f"Test ROC-AUC {roc_auc_score(y_test, p_test):.4f}, accuracy {accuracy_score(y_test, p_test >= 0.5):.4f}")
print("High-income share:", y.mean().round(3))
```
