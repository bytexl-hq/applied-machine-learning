## Setup

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.metrics import confusion_matrix, accuracy_score, precision_score, recall_score, f1_score

raw = pd.read_csv("data/loan_applications.csv")
df = raw.copy()

# Fill missing values: most frequent value for categorical and term columns, median for loan amount
for col in ["Gender", "Married", "Dependents", "Self_Employed", "Loan_Amount_Term", "Credit_History"]:
    df[col] = df[col].fillna(df[col].mode()[0])
df["LoanAmount"] = df["LoanAmount"].fillna(df["LoanAmount"].median())

# Incomes and loan amounts are heavily skewed, so use their logs
df["log_total_income"] = np.log(df["ApplicantIncome"] + df["CoapplicantIncome"])
df["log_loan_amount"] = np.log(df["LoanAmount"])
y = (df["Loan_Status"] == "Y").astype(int).values

X = pd.get_dummies(df[["Gender", "Married", "Dependents", "Education", "Self_Employed", "Property_Area",
                       "Credit_History", "log_total_income", "log_loan_amount", "Loan_Amount_Term"]],
                   drop_first=True, dtype=float)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)

# Standardise only the numeric columns; 0/1 columns stay as they are
numeric = ["log_total_income", "log_loan_amount", "Loan_Amount_Term"]
mu, sd = X_train[numeric].mean(), X_train[numeric].std()
X_train, X_test = X_train.copy(), X_test.copy()
X_train[numeric] = (X_train[numeric] - mu) / sd
X_test[numeric] = (X_test[numeric] - mu) / sd

print(X_train.shape, X_test.shape, "approval rate:", y.mean().round(3))
print(list(X_train.columns))
```
