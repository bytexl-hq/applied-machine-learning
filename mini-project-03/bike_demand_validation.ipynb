```python
import time
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import loguniform, randint
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, PolynomialFeatures
from sklearn.linear_model import LinearRegression, Ridge, Lasso, LogisticRegression
from sklearn.model_selection import (train_test_split, KFold, StratifiedKFold, cross_val_score,
                                     GridSearchCV, RandomizedSearchCV)
from sklearn.metrics import f1_score
from sklearn.exceptions import ConvergenceWarning
warnings.filterwarnings("ignore", category=ConvergenceWarning)

df = pd.read_csv("data/hour.csv", parse_dates=["dteday"])

# level: average hourly rides over the previous 28 days (needs at least 7 days of history)
daily_mean = df.groupby("dteday")["cnt"].mean()
df["level"] = df["dteday"].map(daily_mean.shift(1).rolling(28, min_periods=7).mean())
df = df.dropna(subset=["level"]).reset_index(drop=True)

FEATURES = ["season", "mnth", "hr", "holiday", "weekday", "workingday", "weathersit",
            "temp", "atemp", "hum", "windspeed"]

data_2011 = df[df["yr"] == 0].reset_index(drop=True)
data_2012 = df[df["yr"] == 1].reset_index(drop=True)

idx = np.random.default_rng(42).permutation(len(data_2011))
pilot = data_2011.iloc[idx[:800]].reset_index(drop=True)        # small sample used in Tasks 1-3
rest_2011 = data_2011.iloc[idx[800:]].reset_index(drop=True)    # the rest of 2011, used as validation data

def rmse(y, p):
    return float(np.sqrt(np.mean((np.asarray(y) - np.asarray(p)) ** 2)))

print("2011:", data_2011.shape, " 2012:", data_2012.shape, " pilot:", pilot.shape, " rest of 2011:", rest_2011.shape)
```
