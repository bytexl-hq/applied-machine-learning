```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import linkage, dendrogram
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans, DBSCAN, HDBSCAN, AgglomerativeClustering
from sklearn.mixture import GaussianMixture
from sklearn.neighbors import NearestNeighbors
from sklearn.metrics import silhouette_score, adjusted_rand_score

raw = pd.read_csv("data/marketing_campaign.csv")
SPEND = ["MntWines", "MntFruits", "MntMeatProducts", "MntFishProducts", "MntSweetProducts", "MntGoldProds"]
BUY = ["NumWebPurchases", "NumCatalogPurchases", "NumStorePurchases"]

def plot_clusters(Z, labels, ax, title):
    """Scatter of the first two principal components coloured by cluster (-1 = noise, shown in grey)."""
    xy = PCA(2, random_state=0).fit_transform(Z)
    noise = labels == -1
    ax.scatter(xy[noise, 0], xy[noise, 1], c="lightgrey", s=8, label="noise" if noise.any() else None)
    ax.scatter(xy[~noise, 0], xy[~noise, 1], c=labels[~noise], cmap="tab10", s=8)
    ax.set_title(title); ax.set_xlabel("PC1"); ax.set_ylabel("PC2")

print(raw.shape)
raw[["Year_Birth", "Income", "Dt_Customer", "Recency"] + SPEND[:2] + BUY[:2]].head()
```
