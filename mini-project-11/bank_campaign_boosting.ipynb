```python
import time
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, DecisionTreeRegressor
from sklearn.ensemble import AdaBoostClassifier, BaggingClassifier, GradientBoostingClassifier
from sklearn.metrics import roc_auc_score, average_precision_score, log_loss
import xgboost as xgb
import lightgbm as lgb
from catboost import CatBoostClassifier
import warnings
warnings.filterwarnings("ignore", message=".*eval_set.* is deprecated")

bank = pd.read_csv("data/bank-additional-full.csv", sep=";").drop(columns="duration")
bank["y"] = (bank["y"] == "yes").astype(int)
CATEGORICAL = [c for c in bank.columns if bank[c].dtype == object or str(bank[c].dtype).startswith("str")]
NUMERIC = [c for c in bank.columns if c not in CATEGORICAL + ["y"]]

train_df, test_df = train_test_split(bank, test_size=0.2, stratify=bank["y"], random_state=42)
train_df, valid_df = train_test_split(train_df, test_size=0.25, stratify=train_df["y"], random_state=42)
y_train, y_valid, y_test = train_df["y"].values, valid_df["y"].values, test_df["y"].values

# One-hot version for scikit-learn models
onehot = pd.get_dummies(bank.drop(columns="y"), columns=CATEGORICAL, dtype=float)
X_train, X_valid, X_test = onehot.loc[train_df.index], onehot.loc[valid_df.index], onehot.loc[test_df.index]

def scores(p, y=y_test):
    return {"ROC-AUC": roc_auc_score(y, p), "PR-AUC": average_precision_score(y, p), "log loss": log_loss(y, p)}

print("train/valid/test:", len(train_df), len(valid_df), len(test_df), "| subscription rate:", round(bank["y"].mean(), 3))
print("categorical:", CATEGORICAL)
```
