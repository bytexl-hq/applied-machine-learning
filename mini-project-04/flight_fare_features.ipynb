## Setup

```python
import re
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split, KFold
from sklearn.linear_model import Ridge, Lasso
from sklearn.preprocessing import StandardScaler, PowerTransformer
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.feature_selection import SelectKBest, f_regression, RFE
from sklearn.metrics import mean_absolute_error
from sklearn.exceptions import ConvergenceWarning
warnings.filterwarnings("ignore", category=ConvergenceWarning)

raw = pd.read_csv("data/flight_fares.csv")

# Simulated feed outage: Duration is blank for 10% of rows (fixed seed)
df = raw.copy()
outage = np.random.default_rng(7).random(len(df)) < 0.10
df.loc[outage, "Duration"] = np.nan

def evaluate(X_train, X_test, y_train, y_test, alpha=1.0):
    """Standardises the features, fits Ridge on log(price) and returns
    (5-fold CV MAE on the training set, test MAE), both in rupees."""
    X_train, X_test = np.asarray(X_train, dtype=float), np.asarray(X_test, dtype=float)
    cv_errors = []
    for tr, va in KFold(5, shuffle=True, random_state=0).split(X_train):
        sc = StandardScaler().fit(X_train[tr])
        m = Ridge(alpha=alpha).fit(sc.transform(X_train[tr]), np.log(y_train[tr]))
        cv_errors.append(mean_absolute_error(y_train[va], np.exp(m.predict(sc.transform(X_train[va])))))
    sc = StandardScaler().fit(X_train)
    m = Ridge(alpha=alpha).fit(sc.transform(X_train), np.log(y_train))
    return float(np.mean(cv_errors)), float(mean_absolute_error(y_test, np.exp(m.predict(sc.transform(X_test)))))

print(df.shape, "| missing Duration:", df["Duration"].isna().sum())
df.head()
```
