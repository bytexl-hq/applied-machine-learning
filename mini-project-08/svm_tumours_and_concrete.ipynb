```python
import time
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split, GridSearchCV, StratifiedKFold, KFold
from sklearn.preprocessing import StandardScaler, PolynomialFeatures
from sklearn.pipeline import make_pipeline
from sklearn.svm import SVC, SVR
from sklearn.linear_model import LinearRegression
from sklearn.metrics import accuracy_score, recall_score, mean_absolute_error, r2_score

cancer = pd.read_csv("data/breast_cancer.csv")
cancer["malignant"] = (cancer["diagnosis"] == "malignant").astype(int)
CANCER_FEATURES = [c for c in cancer.columns if c not in ("diagnosis", "malignant")]
C_train, C_test = train_test_split(cancer, test_size=0.3, stratify=cancer["malignant"], random_state=42)
yc_train, yc_test = C_train["malignant"].values, C_test["malignant"].values

concrete = pd.read_csv("data/concrete.csv")
CONCRETE_FEATURES = [c for c in concrete.columns if c != "strength"]
K_train, K_test = train_test_split(concrete, test_size=0.25, random_state=42)
yk_train, yk_test = K_train["strength"].values, K_test["strength"].values

def plot_boundary(model, X, y, ax, title, show_margin=True):
    """Draws the decision function of a fitted 2-feature classifier, its margins and support vectors."""
    xx, yy = np.meshgrid(np.linspace(X[:, 0].min() - 0.5, X[:, 0].max() + 0.5, 300),
                         np.linspace(X[:, 1].min() - 0.5, X[:, 1].max() + 0.5, 300))
    z = model.decision_function(np.c_[xx.ravel(), yy.ravel()]).reshape(xx.shape)
    ax.contourf(xx, yy, z > 0, alpha=0.15, cmap="coolwarm")
    ax.contour(xx, yy, z, levels=[-1, 0, 1] if show_margin else [0], linestyles=["--", "-", "--"] if show_margin else ["-"], colors="k")
    ax.scatter(X[:, 0], X[:, 1], c=y, cmap="coolwarm", s=12, edgecolors="none")
    sv = model.support_vectors_ if hasattr(model, "support_vectors_") else model[-1].support_vectors_
    ax.scatter(sv[:, 0], sv[:, 1], s=60, facecolors="none", edgecolors="k", linewidths=0.8)
    ax.set_title(title)

print("cancer:", C_train.shape, C_test.shape, "| malignant share:", round(cancer["malignant"].mean(), 3))
print("concrete:", K_train.shape, K_test.shape)
```
