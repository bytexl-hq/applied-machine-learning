```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split, GridSearchCV, StratifiedKFold, KFold, cross_val_score
from sklearn.tree import DecisionTreeClassifier, DecisionTreeRegressor, plot_tree, export_text
from sklearn.linear_model import LinearRegression
from sklearn.metrics import accuracy_score, roc_auc_score, recall_score, mean_absolute_error, r2_score

credit = pd.read_csv("data/german_credit.csv")
credit["bad"] = 1 - credit["credit_risk"]          # 1 = the loan went bad
X = pd.get_dummies(credit.drop(columns=["credit_risk", "bad"]), dtype=float)
FEATURES = list(X.columns)
X_train, X_test, y_train, y_test = train_test_split(X, credit["bad"].values, test_size=0.3,
                                                    stratify=credit["bad"], random_state=42)

cars = pd.read_csv("data/auto_mpg.csv")
cars["horsepower"] = cars["horsepower"].fillna(cars["horsepower"].median())
XI = pd.get_dummies(cars.drop(columns=["mpg", "name"]), columns=["origin"], dtype=float)
XI_train, XI_test, yi_train, yi_test = train_test_split(XI, cars["mpg"].values, test_size=0.2, random_state=42)

print("credit:", X_train.shape, X_test.shape, "| bad-loan share:", credit["bad"].mean())
print("cars:", XI_train.shape, XI_test.shape)
```
